### 4 Hyperparameters and Sample Efficiency
Experiment 4: InvertedPendulum

```python
pendulum_inital = [f for f in run_folders if 'pendulum_default' in f]
pendulum_search = [f for f in run_folders if 'pendulum_discount' in f]
pendulum_all = [f for f in run_folders if 'pendulum' in f]
```

actor_info/Actor_Loss

```python
plt.figure(figsize=(10, 8))

for run_folder in pendulum_all:
    event_acc = EventAccumulator(run_folder)
    event_acc.Reload()

    try:
        steps = np.array([int(s.step) for s in event_acc.Scalars('Train_EnvstepsSoFar')])
        values = np.array([s.value for s in event_acc.Scalars('actor_info/Actor_Loss')])

        # Plotting
        plt.plot(steps, values, label=os.path.basename(run_folder))  # using run folder name as label
    except Exception as e:
        print(f"Could not retrieve data for {run_folder}: {e}")

# Adding legends, labels and title
plt.legend(loc='upper center', bbox_to_anchor=(0.5, -0.1), ncol=2)
plt.xlabel('Train_EnvstepsSoFar')
plt.ylabel('actor_info/Actor_Loss')
plt.title('Learning Curve for actor_info/Actor_Loss')

# Show plot
plt.show()
```

```python
plt.figure(figsize=(10, 8))

for run_folder in pendulum_all:
    event_acc = EventAccumulator(run_folder)
    event_acc.Reload()

    try:
        steps = np.array([int(s.step) for s in event_acc.Scalars('Train_EnvstepsSoFar')])
        values = np.array([s.value for s in event_acc.Scalars('critic_info/Baseline_Loss')])

        # Plotting
        plt.plot(steps, values, label=os.path.basename(run_folder))  # using run folder name as label
    except Exception as e:
        print(f"Could not retrieve data for {run_folder}: {e}")

# Adding legends, labels and title
plt.legend(loc='upper center', bbox_to_anchor=(0.5, -0.1), ncol=2)
plt.xlabel('Train_EnvstepsSoFar')
plt.ylabel('critic_info/Baseline_Loss')
plt.title('Learning Curve for critic_info/Baseline_Loss')

# Show plot
plt.show()
```

Eval_AverageReturn

```python
plt.figure(figsize=(10, 8))

for run_folder in pendulum_all:
    event_acc = EventAccumulator(run_folder)
    event_acc.Reload()

    try:
        steps = np.array([int(s.step) for s in event_acc.Scalars('Train_EnvstepsSoFar')])
        values = np.array([s.value for s in event_acc.Scalars('Eval_AverageReturn')])

        # Plotting
        plt.plot(steps, values, label=os.path.basename(run_folder))  # using run folder name as label
    except Exception as e:
        print(f"Could not retrieve data for {run_folder}: {e}")

# Adding legends, labels and title
plt.legend(loc='upper center', bbox_to_anchor=(0.5, -0.1), ncol=2)
plt.xlabel('Train_EnvstepsSoFar')
plt.ylabel('Eval_AverageReturn')
plt.title('Learning Curve for Eval_AverageReturn')

# Show plot
plt.show()
```

```python
inital_returns = []
for run_folder in pendulum_inital:
    event_acc = EventAccumulator(run_folder)
    event_acc.Reload()
    return_values = np.array([s.value for s in event_acc.Scalars('Eval_AverageReturn')])
    inital_returns.append(return_values)

average_returns = np.mean(inital_returns, axis=0)
```

```python
hyp_returns = []
for run_folder in pendulum_search:
    event_acc = EventAccumulator(run_folder)
    event_acc.Reload()
    return_values = np.array([s.value for s in event_acc.Scalars('Eval_AverageReturn')])
    hyp_returns.append(return_values)

hyp_average_returns = np.mean(hyp_returns, axis=0)
```

```python
for f in run_folders:
    if ('default_s1' in f):
        default_s1 = f
    elif('hyp_search_s1' in f):
        hyp_search_s1 = f
```

```python
event_acc1 = EventAccumulator(default_s1)
event_acc1.Reload()
steps = np.array([int(s.step) for s in event_acc1.Scalars('Train_EnvstepsSoFar')])

event_hyp_search1 = EventAccumulator(hyp_search_s1)
event_hyp_search1.Reload()
hyp_steps = np.array([int(s.step) for s in event_hyp_search1.Scalars('Train_EnvstepsSoFar')])
```

```python
plt.figure(figsize=(10, 8))

plt.plot(steps, average_returns, label='Eval_AverageReturn(inital parameter)')
plt.plot(hyp_steps, hyp_average_returns, label='Eval_AverageReturn(search parameter)')
plt.legend(loc='upper center', bbox_to_anchor=(0.5, -0.1), ncol=2)
plt.xlabel('iteration')
plt.ylabel('Eval_AverageReturn')
plt.title('Eval_AverageReturn')

plt.show()
```
